```python
import os
import sys

import django


def load_django_env():
    sys.path.append("../")
    os.environ.setdefault("DJANGO_SETTINGS_MODULE", "config.settings.production")

    django.setup()


load_django_env()
```

```python
import json
import logging

import requests

from config.settings.base import env

API_KEY = env("PK_API_KEY")

logger = logging.getLogger(__name__)

BASE_URL = "http://localhost:8000/api/v1/professions/"


def search_resource(resource: str, search_fields: dict):
    """
    Perform a GET request to the advocates API with dynamic search fields.

    Args:
        search_fields (dict): e.g.,
            {
                "name": "jeckonia",
                "advocate_number": "P.105/11902/15",
                "law_firm": "CMS Kenya | Daly Inamdar Advocates LLP"
            }

    Returns:
        dict: JSON response from the API
    """
    headers = {"PK-Api-Key": API_KEY}
    payload = {"search_fields": search_fields}
    params = {"search": json.dumps(payload)}
    response = requests.get(
        BASE_URL + resource, headers=headers, params=params, timeout=30
    )
    response.raise_for_status()

    return response.json()


advocate_data = search_resource(
    "advocates",
    {
        "name": "jeckonia",
        "advocate_number": "P.105/11902/15",
        "law_firm": "CMS Kenya | Daly Inamdar Advocates LLP",
    },
)

logger.info(advocate_data)
```
